```python
import  numpy as np
import pandas as pd
```

```python
df =  pd.read_csv("/run/media/darshchouhan/New Volume/ML_Algorithm/DATA SETS/placement.csv")
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['cgpa'],df['package'])
plt.xlabel("we")
```

```python
X=  df[['cgpa']]
Y= df['package']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.3,random_state=42)
```

```python
print(Y_train)
```

```python
from  sklearn.linear_model import LinearRegression
lr =  LinearRegression()
model  =  lr.fit(X_train,Y_train)
sol = model.predict(X_test)
from sklearn.metrics import r2_score
```

```python
score= r2_score(sol,Y_test)
print(score)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
model = lr.fit(X_train,Y_train)
```

```python
out = model.predict(X_test)
```

```python
mae = mean_absolute_error(out,Y_test)
print(mae)
```

```python
mAe = mean_squared_error(out,Y_test)
print(mAe)
```

```python
r2 = r2_score(out,Y_test)
print(r2)
```

```python
# Adjusted R2
adjr_2 = 1-((1-r2)*(40-1)/(40-1-1))
```
