```python
import  numpy as np
import pandas as pd
```

```python
df =  pd.read_csv("/run/media/darshchouhan/New Volume/ML_Algorithm/DATA SETS/placement.csv")
```

```python
X=  df[['cgpa']]
Y= df['package']
```

```python

from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.3,random_state=42)
```

```python
class my_LR:
    def __init__(self):
        self.m = None
        self.n = None
    def fit(self,X_train,Y_train):
        X_train =  X_train.iloc[:,0]
      
        num = 0
        den = 0
        for i in range(X_train.shape[0]):
            num = num+(X_train.iloc[i]-X_train.mean())*(Y_train.iloc[i]-Y_train.mean())
            den = den + (X_train.iloc[i]-X_train.mean())*(X_train.iloc[i]-X_train.mean())  
        self.m =num/den
        self.b = (Y_train.mean()-(self.m*X_train.mean()))
        print('Intercept')
        print(self.b)
        print('Slope')
        print(self.m)
            
    def predict(self,X_test):
        X_test = X_test.iloc[:,0]
        return self.m*X_test+self.b
        
        
```

```python
my_model = my_LR()
model = my_model.fit(X_train,Y_train)
```
