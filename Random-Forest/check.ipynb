```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
 
```

```python

X,y = make_circles(n_samples=500,factor=0.1,noise=0.55,random_state=42)
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25)
```

```python
X.shape
```

```python
y.shape
```

```python
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
from sklearn.tree import DecisionTreeClassifier
dtc = DecisionTreeClassifier()
dtc.fit(X_train,y_train)
```

```python
from sklearn.inspection import DecisionBoundaryDisplay

display = DecisionBoundaryDisplay.from_estimator(dtc,X, alpha=0.4)

plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',edgecolors='black')

```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(n_estimators=1000)
rf.fit(X_train,y_train)
```

```python
res1 = dtc.predict(X_test)
res2 = rf.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,res1))
print(accuracy_score(y_test,res2))
```

```python
from sklearn.inspection import DecisionBoundaryDisplay

display = DecisionBoundaryDisplay.from_estimator(rf,X_test,cmap='Set3',alpha=0.9)
plt.scatter(X_train[:,0],X_train[:,1],c=y_train)
```

```python


n_train = 150        
n_test = 1000       
noise = 0.1

# Generate data
def f(x):
    x = x.ravel()
    return np.exp(-x ** 2) + 1.5 * np.exp(-(x - 2) ** 2)

def generate(n_samples, noise):
    X = np.random.rand(n_samples) * 10 - 5
    X = np.sort(X).ravel()
    y = np.exp(-X ** 2) + 1.5 * np.exp(-(X - 2) ** 2)\
        + np.random.normal(0.0, noise, n_samples)
    X = X.reshape((n_samples, 1))

    return X, y

X_train, y_train = generate(n_samples=n_train, noise=noise)
X_test, y_test = generate(n_samples=n_test, noise=noise)
```

```python

plt.figure(figsize=(18, 8))
plt.plot(X_test, f(X_test), "r")
plt.scatter(X_train, y_train, c="b", s=20)
plt.xlim([-5, 5])
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
tr = DecisionTreeRegressor()
tr.fit(X_train,y_train)
```

```python
X_train.shape
```

```python
from sklearn.inspection import DecisionBoundaryDisplay
predict = tr.predict(X_test)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(18, 8))
plt.plot(X_test, f(X_test), "r")
plt.scatter(X_train, y_train)
plt.plot(X_test,predict)
```

## Training the Random Forest

```python
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(n_estimators=2000)
```

```python
rf.fit(X_train,y_train)
```

```python
prediction = rf.predict(X_test)
```
