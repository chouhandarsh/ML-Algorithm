```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/home/darshchouhan/.cache/kagglehub/datasets/arezaei81/heartcsv/versions/1/heart.csv")
```

```python
df.head(5)
```

```python
X = df.drop('target',axis =1)
y = df['target']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.24)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
model1 = LogisticRegression()
model1.fit(X_train,y_train)
model2  = DecisionTreeClassifier()
model2.fit(X_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
```

```python
res1= model1.predict(X_test)
res2 = model2.predict(X_test)
print(accuracy_score(y_test,res1))
print(accuracy_score(y_test,res2))
```

## HOW MUCH ACCURACY IS GOOD

## Confusion Metrices

```python
from sklearn.metrics import confusion_matrix
```

```python
confusion_matrix(y_test,res1)
```

### Type 1

- Number of false positive is called the type 1 error


### Type 2

- False-Negative is called Type-II error

## For Multi-classification Problem

```python
print("Logistic Regression Confusion Metrices")
pd.DataFrame(confusion_matrix(y_test,res1))
```

## Precision & Recall

- Precision Type 1 error is more dangerous
- Recall Type 2 error is more dangerous

# F1 Score $\frac{2PR}{P+R}$

- The main property of harmonic mean is that  is keep the data in the lower side and it pinelises the model for  it
- if P=60, R=100 then the F1 score is $\frac{2*100*60}{160} ~ 70 $ to the lower side

```python
from sklearn.metrics import recall_score,precision_score,f1_score
```

```python
print("For Logistic Regression Model")
print("-"*30)
cdf=pd.DataFrame(confusion_matrix(y_test,res1))
print(cdf)
print("precision",precision_score(y_test ,res1))
print("recall",recall_score(y_test,res1))
print("F1",f1_score(y_test,res1))
```

```python
print("For Desision Tree  Model")
print("-"*30)
cdf=pd.DataFrame(confusion_matrix(y_test,res2))
print(cdf)
print("precision",precision_score(y_test ,res2))
print("recall",recall_score(y_test,res2))
print("F1",f1_score(y_test,res2))
```
