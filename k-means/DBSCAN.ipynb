## Density Based Spatial Clustering of Application with Noise 

- In this we try to find the density of the points to check if it the points surrounding it satisfies the given Hyperpameters  then we can say that it is in the cluster

- In this we use to select a core point a core point is a point such its in its Epsilon neighbour we have min number of datapoints

- Border Point is a point such that from epsilon distance form it we dont have min number of point but there is atleast one core point then it is called border point

- If a Data point is neither the core nor the border then it is called the noise point

## Density Connected Points
- If two points A and B are Density Connected then we can add them inside the same Collection point
- If two points A and B are connected by the core points then it is called the desitity connected points


## DBSCAN
- Step 0 first we decide the value of epsilon and k
- After it we start with the classifying the points as the Border point core point and the noise point
- find the point that is not in the cluster then find the point surrounding it and create a density connected points to the current points into the cluster
- add all the points that are unclustered and desity connected points to the current point into this cluster
- For Each Unclustered border point assign it to the cluster of the nearest core points 
- Leave all the noise points as it is

## Advantages
- Robust to  outliers so it is used for anamoli detection
- No Need to specify the  cluster
- we dont to explicitly mention how many clusters is needed
- Only two Hyperparameters 

## Disadvantage
- It is very sensitive on its Hyperparameters
- Dont perform well when the density are different 
- Dosent do prediction
