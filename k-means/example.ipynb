```python
import  matplotlib.pyplot as plt
plt.scatter(df['cgpa'],df['iq'])
```

```python
from sklearn.cluster import KMeans
```

```python
wcss=[]
for i in range(1,11):
    km=KMeans(n_clusters=i)
    km.fit_predict(df)
    wcss.append(km.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1,11),wcss)
```

```python
X = df.iloc[:,:]
```

```python
km = KMeans(n_clusters=4)
y_means=km.fit_predict(X)
```

```python
plt.scatter(X.iloc[y_means==0,0],X.iloc[y_means==0,1],color='blue')
plt.scatter(X.iloc[y_means==1,0],X.iloc[y_means==1,1],color='red')
plt.scatter(X.iloc[y_means==2,0],X.iloc[y_means==2,1],color='green')
plt.scatter(X.iloc[y_means==3,0],X.iloc[y_means==3,1])
```
