```python
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from sklearn.tree import DecisionTreeClassifier
from sklearn.inspection import DecisionBoundaryDisplay
from itertools import combinations

# Generate all feature pairs
pairs = list(combinations(range(X_train.shape[1]), 2))

plt.figure(figsize=(15, 10))

for i, pair in enumerate(pairs):

    # Select only two features
    X_pair = X_train[:, pair]

    # Train a temporary tree on these two features
    clf = DecisionTreeClassifier(
        criterion="gini",
        random_state=42
    )

    clf.fit(X_pair, y_train)

    ax = plt.subplot(
        (len(pairs)+2)//3,
        3,
        i+1
    )

    disp = DecisionBoundaryDisplay.from_estimator(
        clf,
        X_pair,
        response_method="predict",
        alpha=0.4,
        ax=ax
    )

    ax.scatter(
        X_pair[:,0],
        X_pair[:,1],
        c=y_train,
        cmap=ListedColormap(disp.multiclass_colors_),
        edgecolor="black",
        s=20
    )

    ax.set_xlabel(f"Feature {pair[0]}")
    ax.set_ylabel(f"Feature {pair[1]}")

plt.tight_layout()
plt.show()
```
