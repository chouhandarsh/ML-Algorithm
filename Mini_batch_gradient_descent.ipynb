```python
import numpy as np

class MiniBatchGDRegressor:

    def __init__(self, batch_size=32, learning_rate=0.01, epochs=100):
        self.batch_size = batch_size
        self.lr = learning_rate
        self.epochs = epochs
        self.coef_ = None
        self.intercept_ = None

    def fit(self, X, y):

        # Initialize parameters
        self.coef_ = np.ones(X.shape[1])
        self.intercept_ = 0

        n_samples = X.shape[0]

        for epoch in range(self.epochs):

            # Shuffle data at the beginning of each epoch
            indices = np.random.permutation(n_samples)
            X_shuffled = X[indices]
            y_shuffled = y[indices]

            # Create mini-batches
            for i in range(0, n_samples, self.batch_size):

                X_batch = X_shuffled[i:i+self.batch_size]
                y_batch = y_shuffled[i:i+self.batch_size]

                # Predictions
                y_pred = np.dot(X_batch, self.coef_) + self.intercept_

                # Gradients
                d_coef = (-2 / len(X_batch)) * np.dot(
                    X_batch.T, (y_batch - y_pred)
                )

                d_intercept = (-2 / len(X_batch)) * np.sum(
                    y_batch - y_pred
                )

                # Update parameters
                self.coef_ = self.coef_ - self.lr * d_coef
                self.intercept_ = self.intercept_ - self.lr * d_intercept

    def predict(self, X):
        return np.dot(X, self.coef_) + self.intercept_
```
