```python
import numpy as np
import pandas as pd
from  mlxtend.plotting import plot_decision_regions
```

```python
df = pd.DataFrame()
df['X1'] = [1,2,3,4,5,6,6,7,9,9]    
df['X2'] = [5,3,6,8,1,9,5,8,9,2]
df['label'] = [1,1,0,1,0,1,0,1,0,0]
```

```python
df
```

```python
import seaborn as sns
sns.scatterplot(x=df['X1'],y=df['X2'],hue=df['label'])
```

```python
df['Weights'] = 1/df.shape[0]
```

```python
df
```

```python
from sklearn.tree import DecisionTreeClassifier
dct = DecisionTreeClassifier(max_depth=1)
```

```python
X=df.drop(['label','Weights'],axis=1)
y = df['label']
```

```python
dct.fit(X,y)
```

```python
from sklearn.tree import plot_tree
plot_tree(dct)
```

```python
X = X.to_numpy()
y = y.to_numpy()
```

```python
X
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X,y,dct,legend=2)
```

```python
df['prediction'] = dct.predict(X)
```

```python
def error_finder(df):
    error=0
    for i in range(0,df.shape[0]):
        if (df.iloc[i]['label']!=df.iloc[i]['prediction']):
            error+=df.iloc[i]['Weights']
    return error

    
```

```python
def calculate_model_weight(df):
    error = error_finder(df)
    print("Error",error)
    return 0.5*np.log((1-error)/error)
```

```python
model_weigth = calculate_model_weight(df)
```

```python
model_weigth
```

```python
df
```

```python
# Step 4 - Update weights
def update_row_weights(df,alpha=0.423):
  if df['label'] == df['prediction']:
    return df['Weights'] * np.exp(-alpha)
  else:
    return df['Weights'] * np.exp(alpha)
     
```

```python
df['Updated_weight'] =df.apply(update_row_weights,axis=1)
```

```python
df
```

## Now we will have to normalize the weights

```python
df['Normalized_weight']=df['Updated_weight']/df['Updated_weight'].sum()
```

```python
df
```

```python
df['Normalized_weight'].sum()
```

```python
# After this we will create the weights and then we  will     create the range of them then 
df['cumsum_upper'] = np.cumsum(df['Normalized_weight'])
```

```python
df['cumsum_lower'] = df['cumsum_upper'] - df['Normalized_weight']
```

```python

def create_new_dataset(df):

  indices = []

  for i in range(df.shape[0]):
    a = np.random.random()
    for index,row in df.iterrows():
      if row['cumsum_upper'] > a and a > row['cumsum_lower']:
        indices.append(index)
```
