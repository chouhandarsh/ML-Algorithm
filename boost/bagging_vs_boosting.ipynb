# Bagging vs Boosting

Both **Bagging (Bootstrap Aggregating)** and **Boosting** are **ensemble learning techniques**. They combine multiple weak learners (usually Decision Trees) to build a stronger predictive model.

---

# Bagging

## Definition

Bagging trains **multiple models independently** on different random subsets of the training data and combines their predictions.

Its main goal is to **reduce variance** and improve the stability of the model.

---

## Workflow

```text
                Original Dataset
                       │
        ┌──────────────┼──────────────┐
        │              │              │
 Bootstrap 1     Bootstrap 2    Bootstrap 3
        │              │              │
      Tree 1        Tree 2        Tree 3
        │              │              │
        └────── Majority Vote ────────┘
```

Each model is trained **independently**.

---

## How Bagging Works

### Step 1

Generate multiple **bootstrap samples**.

A bootstrap sample is created by randomly selecting observations **with replacement**.

Example:

Original Dataset

```text
1 2 3 4 5
```

Bootstrap Sample

```text
2 5 5 1 4
```

Notice that:

- Sample **5** appears twice.
- Sample **3** is missing.

---

### Step 2

Train one model on each bootstrap sample.

```text
Bootstrap 1 → Tree 1

Bootstrap 2 → Tree 2

Bootstrap 3 → Tree 3
```

---

### Step 3

Combine predictions.

Classification:

```text
Tree1 → Cat

Tree2 → Dog

Tree3 → Cat

Final Prediction = Cat
```

Regression:

```text
Prediction = Average(Tree Predictions)
```

---

## Advantages

- Reduces variance
- Reduces overfitting
- Easy to parallelize
- Robust to noisy data

---

## Disadvantages

- Does not reduce bias significantly
- Requires multiple models

---

## Popular Algorithms

- BaggingClassifier
- Random Forest
- Extra Trees

---

# Boosting

## Definition

Boosting trains models **sequentially**.

Each new model focuses on the mistakes made by the previous models.

Its primary goal is to **reduce bias** while also improving overall performance.

---

## Workflow

```text
              Original Dataset
                     │
                  Tree 1
                     │
      Increase weights of mistakes
                     │
                  Tree 2
                     │
      Increase weights again
                     │
                  Tree 3
                     │
             Weighted Voting
                     │
               Final Prediction
```

---

## How Boosting Works

Initially every sample has the same weight.

Example

| Sample | Weight |
|--------|---------|
| 1 | 0.10 |
| 2 | 0.10 |
| 3 | 0.10 |
| ... | ... |

---

### Train First Weak Learner

Suppose Tree 1 misclassifies Samples 3 and 7.

Increase their weights.

```text
Sample 3 ↑

Sample 7 ↑
```

Correctly classified samples receive lower weights.

---

### Train Second Weak Learner

Tree 2 now pays more attention to Samples 3 and 7.

Suppose Tree 2 fixes Sample 3 but misclassifies Sample 9.

Increase the weight of Sample 9.

---

### Train Third Weak Learner

Tree 3 focuses more on Sample 9.

This process continues until all weak learners are trained.

---

### Final Prediction

Each learner gets a model weight

$$
\alpha_1,\;\alpha_2,\;\alpha_3,\;\ldots
$$

The final prediction is obtained using **weighted voting**.

---

## Advantages

- Reduces bias
- High prediction accuracy
- Learns from mistakes
- Excellent for structured/tabular datasets

---

## Disadvantages

- Sequential training (cannot be easily parallelized)
- Sensitive to noisy data and outliers
- Can overfit if too many estimators are used

---

## Popular Algorithms

- AdaBoost
- Gradient Boosting
- XGBoost
- LightGBM
- CatBoost

---

# Key Difference

| Feature | Bagging | Boosting |
|----------|----------|-----------|
| Training | Independent | Sequential |
| Goal | Reduce Variance | Reduce Bias |
| Data | Bootstrap Sampling | Same data with updated sample weights |
| Sample Weights | Equal | Updated after every iteration |
| Model Weights | Equal Vote | Weighted Vote |
| Parallel Training | Yes | No |
| Sensitive to Noise | Low | High |
| Overfitting | Less likely | More likely |
| Final Prediction | Majority Vote / Average | Weighted Voting |
| Popular Algorithm | Random Forest | AdaBoost, XGBoost |

# Real-Life Analogy

## Bagging

Imagine asking **10 doctors** to diagnose a patient.

Each doctor works independently.

The final diagnosis is based on **majority voting**.

---

## Boosting

Imagine a **teacher** helping a student.

- Teacher 1 teaches the lesson.
- Teacher 2 focuses only on the student's mistakes.
- Teacher 3 focuses on the remaining mistakes.

Each teacher improves upon the previous teacher.

---

# When to Use

### Use Bagging when

- Your model has **high variance**
- The model is overfitting
- You want fast parallel training
- Noise is present in the data

---

### Use Boosting when

- Your model has **high bias**
- You need maximum prediction accuracy
- You can afford longer training time
- The dataset is relatively clean

---

# Summary

| Bagging | Boosting |
|----------|-----------|
| Parallel Training | Sequential Training |
| Independent Models | Dependent Models |
| Bootstrap Sampling | Weight Updating |
| Majority Voting | Weighted Voting |
| Reduces Variance | Reduces Bias |
| Example: Random Forest | Example: AdaBoost, XGBoost |
