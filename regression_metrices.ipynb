# Multiple Linear Regression

## Hypothesis

$$
\hat{y}
=======

\beta_0
+
\beta_1x_1
+
\beta_2x_2
+\cdots+
\beta_mx_m
$$

---

## Matrix Representation

Predicted Vector

$$
\hat{Y}
=======

\begin{pmatrix}
\hat{y}_1 \
\hat{y}_2 \
\hat{y}_3 \
\vdots \
\hat{y}_n
\end{pmatrix}
$$

Feature Matrix

$$
X
=

\begin{pmatrix}
1 & x_{11} & x_{12} & \cdots & x_{1m} \
1 & x_{21} & x_{22} & \cdots & x_{2m} \
1 & x_{31} & x_{32} & \cdots & x_{3m} \
\vdots & \vdots & \vdots & \ddots & \vdots \
1 & x_{n1} & x_{n2} & \cdots & x_{nm}
\end{pmatrix}
$$

Coefficient Vector

$$
\beta
=====

\begin{pmatrix}
\beta_0 \
\beta_1 \
\beta_2 \
\vdots \
\beta_m
\end{pmatrix}
$$

Therefore

$$
\boxed{
\hat{Y}=X\beta
}
$$

---

## Actual Output Vector

$$
Y
=

\begin{pmatrix}
y_1 \
y_2 \
y_3 \
\vdots \
y_n
\end{pmatrix}
$$

---

## Residual Vector

$$
e
=

Y-\hat{Y}
$$

or

$$
e
=

\begin{pmatrix}
y_1-\hat{y}_1 \
y_2-\hat{y}_2 \
y_3-\hat{y}_3 \
\vdots \
y_n-\hat{y}_n
\end{pmatrix}
$$

---

## Sum of Squared Errors

$$
E
=

\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
$$

Vector Form

$$
E=e^{T}e
$$

Substituting

$$
e=Y-\hat{Y}
$$

gives

$$
E=(Y-\hat{Y})^T(Y-\hat{Y})
$$

Since

$$
\hat{Y}=X\beta
$$

we obtain

$$
E=(Y-X\beta)^T(Y-X\beta)
$$

---

## Expansion

$$
E=
[Y^T-(X\beta)^T]
[Y-X\beta]
$$

$$
E=
Y^TY
-
Y^TX\beta
-
(X\beta)^TY
+
(X\beta)^T(X\beta)
$$

## Using

$$
(X\beta)^T
=
\beta^TX^T
$$

$$
(X\beta)^TY
=
\beta^TX^TY
$$

Since

$$
(Y^TX\beta)^T
=
Y^TX\beta
$$

Therefore

$$
E
=
Y^TY
-
2Y^TX\beta
+
\beta^TX^TX\beta
$$

## Differentiation

$$
\frac{\partial E}{\partial \beta}
=================================

-2X^TY
+
2X^TX\beta
$$

Set equal to zero

$$
-2X^TY
+
2X^TX\beta
==========

0
$$

$$
X^TX\beta
=========

X^TY
$$

---

## Normal Equation

Multiplying both sides by

$$
(X^TX)^{-1}
$$

gives

$$
(X^TX)^{-1}(X^TX)\beta
======================

(X^TX)^{-1}X^TY
$$

Therefore

$$
\boxed{
\beta
=====

(X^TX)^{-1}X^TY
}
$$

where

$$
\beta
=====

\begin{pmatrix}
\beta_0 \
\beta_1 \
\beta_2 \
\vdots \
\beta_m
\end{pmatrix}
$$

This is the closed-form solution of Multiple Linear Regression.
