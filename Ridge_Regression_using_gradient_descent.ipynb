# Ridge Regression

## Design Matrix, Target Vector and Weight Vector

$$
X =
\begin{bmatrix}
1 & x_{11} & x_{12} & x_{13} & \cdots & x_{1n} \\
1 & x_{21} & x_{22} & x_{23} & \cdots & x_{2n} \\
\vdots & \vdots & \vdots & \vdots & \ddots & \vdots \\
1 & x_{m1} & x_{m2} & x_{m3} & \cdots & x_{mn}
\end{bmatrix}
$$

$$
Y =
\begin{bmatrix}
y_1 \\
y_2 \\
\vdots \\
y_m
\end{bmatrix}
$$

$$
W =
\begin{bmatrix}
w_0 \\
w_1 \\
w_2 \\
\vdots \\
w_n
\end{bmatrix}
$$

The prediction equation is:

$$
Y = XW
$$

---

## Ridge Regression Cost Function

$$
J(W)
=
(Y-XW)^T(Y-XW)
+
\lambda W^TW
$$

where

- $Y-XW$ is the residual vector
- $\lambda$ is the regularization parameter
- $W^TW$ is the L2 penalty term

---

## Closed Form Solution (Normal Equation)

$$
\hat{W}
=
(X^TX+\lambda I)^{-1}X^TY
$$

where

$$
I =
\begin{bmatrix}
1 & 0 & \cdots & 0 \\
0 & 1 & \cdots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \cdots & 1
\end{bmatrix}
$$

---

## Gradient Vector

$$
\frac{\partial L}{\partial W}
=
\begin{bmatrix}
\dfrac{\partial L}{\partial w_0} \\
\dfrac{\partial L}{\partial w_1} \\
\vdots \\
\dfrac{\partial L}{\partial w_n}
\end{bmatrix}
$$

---

## Gradient Descent Update Rule

$$
W_{\text{new}}
=
W_{\text{old}}
-
\eta
\left(
\frac{\partial L}{\partial W}
\right)
$$

or equivalently

$$
W_{\text{new}}
=
W_{\text{old}}
-
\eta
\begin{bmatrix}
\dfrac{\partial L}{\partial w_0} \\
\dfrac{\partial L}{\partial w_1} \\
\vdots \\
\dfrac{\partial L}{\partial w_n}
\end{bmatrix}
$$

where

- $\eta$ = Learning Rate
- $L$ = Loss Function
- $\frac{\partial L}{\partial W}$ = Gradient Vector

---
$$

$$
## Gradient of Ridge Regression

$$
\frac{\partial J}{\partial W}
=
-\frac{2}{m}X^T(Y-XW)
+
2\lambda W
$$

---
## Derivation of Ridge Regression Gradient

The Ridge Regression loss function is

$$
L(W)
=
(Y-XW)^T(Y-XW)
+
\lambda W^TW
$$

### Step 1: Expand the first term

$$
L(W)
=
Y^TY
-
Y^TXW
-
(XW)^TY
+
(XW)^T(XW)
+
\lambda W^TW
$$

Since

$$
Y^TXW=(XW)^TY
$$

we get

$$
L(W)
=
Y^TY
-
2Y^TXW
+
W^TX^TXW
+
\lambda W^TW
$$

---

### Step 2: Differentiate each term

#### First Term

$$
\frac{\partial}{\partial W}(Y^TY)=0
$$

because it does not contain \(W\).

#### Second Term

$$
\frac{\partial}{\partial W}
\left(
-2Y^TXW
\right)
=
-2X^TY
$$

#### Third Term

$$
\frac{\partial}{\partial W}
\left(
W^TX^TXW
\right)
=
2X^TXW
$$

#### Fourth Term

$$
\frac{\partial}{\partial W}
\left(
\lambda W^TW
\right)
=
2\lambda W
$$

---

### Step 3: Combine all derivatives

$$
\frac{\partial L}{\partial W}
=
0
-
2X^TY
+
2X^TXW
+
2\lambda W
$$

Therefore,

$$
\frac{\partial L}{\partial W}
=
-2X^TY
+
2X^TXW
+
2\lambda W
$$

Taking 2 common,

$$
\frac{\partial L}{\partial W}
=
2
\left(
X^TXW
-
X^TY
+
\lambda W
\right)
$$

---

### Gradient Descent Update Rule

$$
W_{new}
=
W_{old}
-
\eta
\left(
\frac{\partial L}{\partial W}
\right)
$$

Substituting the gradient,

$$
W_{new}
=
W_{old}
-
\eta
\left[
-2X^TY
+
2X^TXW
+
2\lambda W
\right]
$$

or

$$
W_{new}
=
W_{old}
-
2\eta
\left(
X^TXW
-
X^TY
+
\lambda W
\right)
$$

---

### Setting Gradient Equal to Zero

For the optimal solution,

$$
\frac{\partial L}{\partial W}=0
$$

Therefore,

$$
-2X^TY
+
2X^TXW
+
2\lambda W
=
0
$$

Dividing by 2,

$$
-X^TY
+
X^TXW
+
\lambda W
=
0
$$

$$
X^TXW
+
\lambda W
=
X^TY
$$

Factoring \(W\),

$$
(X^TX+\lambda I)W
=
X^TY
$$

Multiplying both sides by

$$
(X^TX+\lambda I)^{-1}
$$

gives

$$
W
=
(X^TX+\lambda I)^{-1}X^TY
$$

which is the closed-form solution of Ridge Regression.
## Ridge Gradient Descent Update

$$
W_{\text{new}}
=
W_{\text{old}}
-
\eta
\left[
-\frac{2}{m}X^T(Y-XW)
+
2\lambda W
\right]
$$

Simplified:

$$
W_{\text{new}}
=
W_{\text{old}}
+
\frac{2\eta}{m}X^T(Y-XW)
-
2\eta\lambda W
$$

## Creating own Gradient Descent

```python
class MyRidgeGD:
    def __init__(self,epochs=100,learning_rate=0.1,alpha=0.1):
        self.learning_rate = learning_rate
        self.epochs=epochs
        self.alpha = alpha
        self.coeff_=None
        self.intercept_= None
    def fit(self,X_train,y_train):
        self.coeff_ = np.ones(X_train.shape[1])
        self.intercept_=0
        # in the coeff_ matrix the first one was the coeff of the so it is 0
        w = np.insert(self.coeff_,0,self.intercept_)
        X_train=np.insert(X_train,0,1,axis=1)
        for i in range(self.epochs):
            w_der = np.dot(X_train.T,X_train).dot(w)-np.dot(X_train.T,y_train)+np.dot(self.alpha,w)
            w = w-self.learning_rate*w_der
        self.intercept_ = w[0]
        self.coeff_ = w[1:]
    def predict(self,X_test):
        return np.dot(X_test,self.coeff_)+self.intercept_


```

```python
reg2 = MyRidgeGD(epochs=500,learning_rate=0.005,alpha=0.001)
reg2.fit(X_train,y_train)
out2 = reg2.predict(X_test)
print(r2_score(y_test,out2))
```
