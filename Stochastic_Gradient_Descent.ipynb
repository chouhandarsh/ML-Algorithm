## The problem with Batch Gradient Descent

take the example of n=1000, col=5, epochs = 50 

so we will have to calulate 6 derivate for each row and we have n = 1000  so 6000 detrivates and epoches is 50 so in total we have to find 30 lakhs derivaties

#### in a btch gradient descent we use to look the whole data at once then we use to updte  the row and the column this is  so this is not computational effective so we use  Stochastic gradient descent in which we check the one row at a  time and update the all the coefficents 

###  Benifits of the Stocastic Gradient descent are
- we dont have to load the complete data into the system
- we get the too early convergence

- as the name the stochastic in this a row is selected randomly for the calculation so it is called stochastic gradient descent

```python
from sklearn.linear_model import LinearRegression
import numpy as np
import pandas as pd
from sklearn.datasets import load_diabetes
X,y = load_diabetes(return_X_y=True)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.metrics import r2_score
pred = model.predict(X_test)
res = r2_score(y_test,pred)
```

```python
class SGDRegressor:
    def __init__(self,learning_rate=0.5,epochs=100):
        self.coeff_ = None
        self.intercept_ = None
        self.lr = learning_rate
        self.epochs= epochs
    def fit(self,X_train,y_train):
        #initialzing the numbers
        self.intercept_ = 0
        self.coeff_ = np.ones(X_train.shape[-1])
        for i in range(self.epochs):
            for j in range(X_train.shape[0]):
                # generating a random integer
                idx = np.random.randint(0,X_train.shape[0])
                y_hat = np.dot(X_train[idx],self.coeff_)+self.intercept_
                intercept_der = -2*(y_train[idx]-y_hat)
                self.intercept_ = self.intercept_ - (self.lr*intercept_der)
                coef_der = (-2/X_train.shape[0]) * np.dot(X_train[idx], (y_train[idx] - y_hat))
                self.coeff_ = self.coeff_ - self.lr*coef_der
            print(self.intercept_)
            print(self.coeff_)
    def predict(self,X_test):
        # from above  we will get the all the betas and the coeff so we will multiply the all the X_test*bitas + intercept
        return np.dot(X_test,self.coeff_)+self.intercept_

```

```python
mod = SGDRegressor(learning_rate=0.15,epochs=800)
```

```python
mod.fit(X_train,y_train)
```

```python
out =mod.predict(X_test)
res=r2_score(y_test,out)
```

### When to use the stochastic gradient descent

- Big Data
- Non convex function (it is local as well as global minima)
- The main issue is that even after  reaching close to the  solution due to variying we still cannot reach find the exact solution
- for that we use  vary the lr by making it the fucntion of epochs
