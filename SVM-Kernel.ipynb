```python
import numpy as np
import pandas as pd
from sklearn.svm import SVC
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from matplotlib.colors import ListedColormap 
```

```python
from sklearn.datasets._samples_generator import make_circles
```

```python
X,y =  make_circles(200,noise=0.1,factor=0.12,random_state=42)
plt.scatter(X[:,0],X[:,1],c=y,cmap='bwr')
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25)
```

```python
classifier = SVC(kernel='linear')
classifier.fit(X_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,classifier.predict(X_test)))
```

```python
from mlxtend.plotting  import plot_decision_regions
plot  = plot_decision_regions(X,y,classifier)
```

```python
def plot_3d(X,y):
    r = np.exp(-(X**2).sum(1))
    ax = plt.subplot(projection='3d')
    ax.scatter(X[:,0],X[:,1],r,c=y,s=200,cmap='bwr')
```

```python
plot_3d(X,y)
```

```python
def plot_2D(X,y):
    r = np.exp(-(X**2))
    plt.scatter(r[:,0],r[:,1],c=y)
    print(r.shape)
```

```python
plot_2D(X,y)
```

```python
X,y =  make_circles(200,noise=0.35,factor=0.1,random_state=42)
plt.scatter(X[:,0],X[:,1],c=y,cmap='bwr')
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25)
```

```python
rbf_classifier = SVC(kernel='rbf')
rbf_classifier.fit(X_train,y_train)
```

```python
print(accuracy_score(y_test,rbf_classifier.predict(X_test)))
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X,y,rbf_classifier)
```

```python
poly_classifier = SVC(kernel='poly',degree=3)
poly_classifier.fit(X_train,y_train)
print(accuracy_score(y_test,poly_classifier.predict(X_test)))
```

```python
from mlxtend.plotting import  plot_decision_regions
plot_decision_regions(X,y,poly_classifier)
```
