## SMOTE
- Synthetic Minority oversampling technique


- What we do is that we take our minority datapoint and train a k nearest neightbour algorithm and find the 5 nearest neighbour of our minority class
- we select a point randomly
- Select a neighbour randomly from that neighbours
- Select a random number between 0-1 
- calculate new Example sample old-sample-factor(original-neighbour)
- Final Data contains the old Data with the  new generated point|


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
import numpy as np
import xgboost as xgb
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.metrics import classification_report, roc_auc_score, roc_curve
from sklearn.model_selection import train_test_split

# Set random seed for reproducibility
np.random.seed(42)

# Generate imbalanced dataset
n_samples_1 = 25  # Number of samples in class 1
n_samples_2 = 375  # Number of samples in class 2
centers = [(0, 0), (2, 2)]  # Centers of each cluster
cluster_std = [1.5, 1.5]  # Standard deviation of each cluster

X, y = make_blobs(n_samples=[n_samples_1, n_samples_2],
                  centers=centers,
                  
                  cluster_std=cluster_std,
                  random_state=0)
plt.figure(figsize=(8, 6))
plt.scatter(X[y == 0][:, 0], X[y == 0][:, 1], color='red', label='Class 1')
plt.scatter(X[y == 1][:, 0], X[y == 1][:, 1], color='blue', label='Class 2')
plt.title('2D Imbalanced Dataset with Two Classes')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from imblearn.over_sampling import SMOTE
sm = SMOTE()
X_resample,y_resample = sm.fit_resample(X,y)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(X_resample[y_resample == 0][:, 0], X_resample[y_resample == 0][:, 1], color='red', label='Class 1')
plt.scatter(X_resample[y_resample == 1][:, 0], X_resample[y_resample == 1][:, 1], color='blue', label='Class 2')
plt.title('2D Imbalanced Dataset with Two Classes')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()

```
